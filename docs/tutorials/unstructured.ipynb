```python
import earthkit.data as ekd

ds = ekd.from_source("sample", "synop_10.bufr")
```

```python
df = ds.to_pandas(columns=["latitude", "longitude", "airTemperatureAt2M"])
df
```

```python
import xarray as xr

ds = xr.Dataset.from_dataframe(df)
ds
```

```python
import earthkit.geo as ekg

r = ekg.regrid(ds, out_grid={"grid": [1, 1]})
```
